# Convert to Parquet

```python
import pandas as pd
```

```python
year = 2021
```

```python
## this should be set to the filepath of the natality stats addendum file
filepath = f"../data/{year}/raw/Nat{year}PS.txt"
```

### Extract Features from FWF


```python
names_and_cols_dict = {
    "mothers_occurrence_territory": (24, 25),
    "occurrence_fips_county": (28, 30),
    "occurrence_county_population": (31, 31),
    "mothers_birth_country": (80, 81),
    "mothers_residence_country": (85, 86),
    "mothers_residence_territory": (89, 90),
    "residence_fips_county": (91, 93),
    "population_of_residence_county": (99, 99),
    "population_of_residence_city": (100, 100),
    "record_type": (103, 103),
}
```

### Read Into Dataframe


```python
chunks = pd.read_fwf(
    filepath,
    colspecs=list(names_and_cols_dict.values()),
    names=list(names_and_cols_dict.keys()),
    converters={n: str for n in names_and_cols_dict},
    header=None,
    chunksize=500000,
)
```

```python
df = pd.concat(chunks)
```

## Write Into Parquet


```python
df.to_parquet(f"../data/{year}/processed/natality_{year}_residence_addendum.parquet")
```
