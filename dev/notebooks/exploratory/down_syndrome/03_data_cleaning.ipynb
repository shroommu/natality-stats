## Down Syndrome Data Cleaning

```python
import pandas as pd
import pickle as pkl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# suppress warnings
import warnings

warnings.filterwarnings("ignore")
```

### Load Data

```python
chunks = pkl.load(open("../../../data/interim/down_syndrome/chunks_filtered.pkl", "rb"))
target_chunks = pkl.load(
    open("../../../data/interim/down_syndrome/target_chunks.pkl", "rb")
)
feature_importances = pkl.load(
    open("../../../data/processed/down_syndrome/feature_importances_df.pkl", "rb")
)
```

```python
df = pd.concat(chunks).reset_index(drop=True)
target = pd.concat(target_chunks).reset_index(drop=True)

df["target"] = target
```

First, we'll drop the "interval since" features due to their large NA ratio.

```python
df = df.drop(
    columns=[
        "interval_since_last_live_birth",
        "interval_since_last_other_pregnancy_recode",
        "interval_since_last_pregnancy_recode",
    ]
)
```

Then we'll drop missing values from the columns that interest us.

```python
df = df.drop(df[df["fathers_combined_age"] == 99].index)
df = df.drop(df[df["prior_other_terminations"] == 99].index)
df = df.drop(df[df["prior_births_now_living"] == 99].index)
df = df.drop(df[df["pre_pregnancy_weight_recode"] == 999].index)
df = df.drop(df[df["mothers_hispanic_origin_recode"] == 999].index)
df = df.drop(df[df["fathers_hispanic_origin"] == 999].index)
```
