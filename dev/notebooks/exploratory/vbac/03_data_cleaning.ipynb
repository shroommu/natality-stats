## VBAC Data Cleaning

```python
import pandas as pd
import pickle as pkl
```

```python
# suppress warnings
import warnings

warnings.filterwarnings("ignore")
```

### Load Data

```python
df = pkl.load(open("../../../data/2021/interim/vbac/filtered_df.pkl", "rb"))
target = pkl.load(open("../../../data/2021/interim/vbac/target.pkl", "rb"))
```

```python
df["target"] = target
```

We'll select the features determined by our data exploration.

```python
df = df[
    [
        "augmentation_of_labor",
        "induction_of_labor",
        "number_of_previous_cesarean",
        "combined_gestation_detail_in_weeks",
        "prior_births_now_living",
        "fetal_presentation_at_delivery",
        "BMI",
        "target",
    ]
]
```

Then we'll drop any remaining missing values, as well as the "diagnosis pending" values from our target.

```python
df = df.drop(df[df["number_of_previous_cesarean"] == 99].index)
df = df.drop(df[df["combined_gestation_detail_in_weeks"] == 99].index)
df = df.drop(df[df["prior_births_now_living"] == 99].index)
df = df.drop(df[df["fetal_presentation_at_delivery"] == 9].index)
df = df.drop(df[df["BMI"] == 99].index)
df = df.dropna()
```

We'll also remove some outliers:
- Deliveries before 28 weeks gestation.

```python
df = df.drop(df[df["combined_gestation_detail_in_weeks"] < 28].index)
```
