## Down Syndrome Data Cleaning

```python
import pandas as pd
import pickle as pkl
```

```python
# suppress warnings
import warnings

warnings.filterwarnings("ignore")
```

### Load Data

```python
chunks = pkl.load(open("../../../data/interim/down_syndrome/chunks_filtered.pkl", "rb"))
target_chunks = pkl.load(
    open("../../../data/interim/down_syndrome/target_chunks.pkl", "rb")
)
```

```python
df = pd.concat(chunks).reset_index(drop=True)
target = pd.concat(target_chunks).reset_index(drop=True)

df["target"] = target
```

We'll select the features determined by our data exploration.

```python
df = df[["mothers_single_year_age", "fathers_combined_age", "target"]]
```

Then we'll drop any remaining missing values, as well as the "diagnosis pending" values from our target.

```python
df = df.drop(df[df["fathers_combined_age"] == 99].index)
df = df.drop(df[df["target"] == 2].index)
df = df.dropna()
```
